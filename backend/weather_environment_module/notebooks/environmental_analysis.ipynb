# Environmental Data Analysis (Demo)

Quick exploratory notebook over the synthetic demo dataset at `data/demo/sample_environmental_data.csv`.
Useful for sanity-checking the rainfall "story arc" (light -> increasing -> persistent -> high accumulation) used in SIH demos, and for prototyping new features before adding them to `app/processing/features.py`.

```python
import pandas as pd

df = pd.read_csv('../data/demo/sample_environmental_data.csv', parse_dates=['timestamp_utc'])
df = df.sort_values('timestamp_utc')
df.head()
```

```python
df['rainfall_24h_mm'] = df['rainfall_mm'].rolling(24, min_periods=1).sum()
df['rainfall_72h_mm'] = df['rainfall_mm'].rolling(72, min_periods=1).sum()
df[['timestamp_utc', 'rainfall_mm', 'rainfall_24h_mm', 'rainfall_72h_mm']].tail(10)
```

```python
ax = df.set_index('timestamp_utc')[['rainfall_mm']].plot(figsize=(12, 4), title='Synthetic hourly rainfall (demo)')
ax.set_ylabel('mm')
```

```python
ax = df.set_index('timestamp_utc')[['rainfall_24h_mm', 'rainfall_72h_mm']].plot(figsize=(12, 4), title='Rolling rainfall accumulation (demo)')
ax.set_ylabel('mm')
```

The rolling-accumulation plot should visibly show low rainfall in the early part of the window, ramping up, then a persistent/heavy phase near the end - matching the DEMO_MODE story arc implemented in `app/providers/demo.py`.
